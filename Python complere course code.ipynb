## 1. Recursive Factorial

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

num = int(input("Enter a number: "))
print("Factorial =", factorial(num))
```

## 2. Fibonacci Series Using Recursion

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

terms = int(input("How many terms? "))

for i in range(terms):
    print(fibonacci(i), end=" ")
```

## 3. Prime Number Function

```python

def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True


number = int(input("Enter a number: "))

if is_prime(number):
    print("Prime Number")
else:
    print("Not Prime Number")
```

## 4. Palindrome Function

```python
def palindrome(text):
    text = text.lower()
    return text == text[::-1]

word = input("Enter a word: ")

if palindrome(word):
    print("Palindrome")
else:
    print("Not Palindrome")
```

## 5. Armstrong Number Function

```python
def armstrong(num):
    digits = len(str(num))
    total = 0
    temp = num

    while temp > 0:
        digit = temp % 10
        total += digit ** digits
        temp //= 10

    return total == num

number = int(input("Enter a number: "))

if armstrong(number):
    print("Armstrong Number")
else:
    print("Not Armstrong")
```

## 6. Count Vowels

```python
def count_vowels(sentence):
    vowels = "aeiouAEIOU"
    count = 0

    for letter in sentence:
        if letter in vowels:
            count += 1

    return count

text = input("Enter a sentence: ")
print("Total vowels:", count_vowels(text))
```

## 8. Password Strength Checker

```python
def check_password(password):

    if len(password) < 8:
        return "Weak Password"

    has_upper = False
    has_lower = False
    has_digit = False

    for ch in password:
        if ch.isupper():
            has_upper = True
        elif ch.islower():
            has_lower = True
        elif ch.isdigit():
            has_digit = True

    if has_upper and has_lower and has_digit:
        return "Strong Password"

    return "Medium Password"


password = input("Enter Password: ")
print(check_password(password))
```

# 9. Bank Account (Using Functions)

```python
balance = 5000

def deposit(amount):
    global balance
    balance += amount
    print("Balance:", balance)

def withdraw(amount):
    global balance

    if amount <= balance:
        balance -= amount
        print("Balance:", balance)
    else:
        print("Insufficient Balance")

deposit(2000)
withdraw(3000)
withdraw(6000)
```

## 10. Mini Calculator

```python
def calculator(a, b, operator):

    if operator == "+":
        return a + b

    elif operator == "-":
        return a - b

    elif operator == "*":
        return a * b

    elif operator == "/":
        if b != 0:
            return a / b
        return "Cannot divide by zero"

    else:
        return "Invalid Operator"


num1 = float(input("First Number: "))
op = input("Operator (+ - * /): ")
num2 = float(input("Second Number: "))

print("Result =", calculator(num1, num2, op))
```
